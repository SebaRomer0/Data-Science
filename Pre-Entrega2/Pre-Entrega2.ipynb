# Integración y normalización de datos de ventas


```python
#Aca importo todas las librerias que necesito

import pandas as pd
import os
```

```python
#Creo el Data Frame de Ventas

ventas = pd.DataFrame({
    "id_producto": [101, 102, 103, 101],
    "cantidad": [2, 1, 5, 3],
    "fecha": ["2026-09-01", "2026-09-01", "2026-09-02", "2026-09-03"]
})

#Aca guardo el DF en CSV

ventas.to_csv("ventas.csv", index=False)

ventas
```

```python
#Creamos el Data frame de Productos

productos = pd.DataFrame({
    "id_producto": [101, 102, 103],
    "nombre_producto": ["Teclado", "Mouse", "Monitor"],
    "precio_unitario": [5000,1500,180000]
})

#Aca guardo el df en formato Excel

productos.to_excel("productos.xlsx", index=False)

productos
```

```python
#Aca uno las dos tablas mediante el indice ID_Producto con el MERGE

ventas_completas = pd.merge(
    ventas,
    productos,
    on="id_producto",
    how="left"
)

ventas_completas
```

```python
ventas_completas["total_venta"] = ventas_completas["cantidad"] * ventas_completas["precio_unitario"]

ventas_completas
```

```python
#Convierto la fecha como DateTime ya que pandas lo interpretaba como Texto

ventas_completas["fecha"] = pd.to_datetime(
    ventas_completas["fecha"]
)

ventas_completas.dtypes
```

```python
#verifico la existencia de Nulos

ventas_completas.isnull().sum()
```

```python
#convierto Ventas Completas en archivo Parquet

ventas_completas.to_parquet(
    "ventas_final.parquet",
    index=False
)
```

```python
#Importo la libreria OS arriba y ejecuto listdir para ver como esta cada archivo guardado.

os.listdir()
```

```python
#Aca leo el archivo Parquet

df_final = pd.read_parquet("ventas_final.parquet")

df_final
```

```python
#Imprimimos cada resultado para una muestra Final

print("Filas:", len(df_final))
print("Columnas:", list(df_final.columns))
print("Valores nulos:")
print(df_final.isnull().sum())
print("\nTipos de datos:")
print(df_final.dtypes)
```
